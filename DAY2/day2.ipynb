```python
# Day 2
import pandas as pd
import numpy as np

# Creating a DataFrame from dictionary
data = {
    'student_name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Bob'],
    'hours_studied': [5, 8, 3, 10, None, 8],
    'previous_score': [65, 85, 55, 92, 75, 85],
    'attendance': [80, 90, 70, 95, 85, 90],
    'sleep_hours': [7.5, 8.0, 6.0, 8.5, 7.0, 8.0],
    'extracurricular': ['Yes', 'No', 'Yes', 'No', 'Yes', 'No'],
    'parent_education': ['Bachelor', 'Master', 'High School', 'PhD', 'Bachelor', 'Master']
}

df = pd.DataFrame(data)
print("Student Performance DataFrame:")
print(df)

print('Data info')
print(df.info())

# Basic DataFrame information
print("\nDataFrame shape:", df.shape[0], "rows,", df.shape[1], "columns")
print("Column names:", df.columns.tolist())
print("Data types:\n", df.dtypes)
print("Summary statistics:\n", df.describe())
```

```python
# View data
print("First 5 rows:\n", df.head(2))
print("Last 5 rows:\n", df.tail(2))

# check for missing values
print("Missing values:\n", df.isnull().sum())

# Information abut dataFrame
print("DataFrame info:")
print(df.info())

# check nan
print("Check for NaN values:\n", df.isna().sum())

# check duplicates
print("Check for duplicate rows:\n", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()
print("DataFrame after removing duplicates:\n", df)
```
